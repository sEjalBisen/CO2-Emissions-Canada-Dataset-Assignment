```python
# Run this code cell
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

mpl.style.use('ggplot')

import warnings
warnings.filterwarnings('ignore')

url = 'https://new-assets.ccbp.in/frontend/content/aiml/classical-ml/co2_emissions_canada.csv'

# Upload the data on colab and execute this code cell
data = pd.read_csv(url)
data.head()
```

---
**Q1. Identify which 5 vehicle classes have the highest average combined fuel consumption.**

```python
top_5_classes = data.groupby('Vehicle Class')['Fuel Consumption Comb (mpg)'].mean().sort_values(ascending=False).head(5)

print(top_5_classes)
```

---
**2. Find the transmission type with the lowest average CO2 emissions.** print it in the given format:
`____ is the lowest CO2 emission average in transmission type.`

```python
best_transmission = data.groupby("Transmission")["CO2 Emissions(g/km)"].mean().idxmin()
print(best_transmission, "is the lowest CO2 emission average in transmission type.")
```

---
Q3. Which are the top 5 car makes have the highest average fuel consumption in city driving (L/100 km) based on this metric?

```python
# write your code here

high_avg_fuel_con =  data.groupby('Make')['Fuel Consumption City (L/100 km)'].mean().head(5)

print(high_avg_fuel_con.sort_values(ascending=False))
```

```python
# Write your code here to create a visualization
plt.figure(figsize=(8, 5))
plt.bar(high_avg_fuel_con.index, high_avg_fuel_con.values, color='salmon')
plt.xlabel('Car Make')
plt.ylabel('Avg. Fuel Consumption in City (L/100 km)')
plt.title('Average City Fuel Consumption of Selected Car Makes')
plt.tight_layout()
plt.show()

```

---
**4. Find which 3 car manufacturers have the highest median CO2 emissions and display both the make and the median value.**

```python
# Write your code here

top_3_median_emissions = (data.groupby('Make')['CO2 Emissions(g/km)']
                          .median()
                          .sort_values(ascending=False)
                          .head(3))
print(top_3_median_emissions)
```

```python
# Write your code here to create a visualization
top_3_median_emissions.plot(kind='pie',
                             autopct='%1.1f%%',
                             startangle=140,
                             ylabel='',  # Hides y-axis label
                             title='Top 3 Car Makes by Median CO₂ Emissions')

plt.axis('equal')  # Ensures the pie is a circle
plt.show()
```

---
**5. Find the car model with the highest CO2 emissions for each manufacturer.**

```python
# Write your code here

highest_co2_per_make = data.loc[data.groupby("Make")["CO2 Emissions(g/km)"].idxmax(), ["Make", "Model", "CO2 Emissions(g/km)"]]
print(highest_co2_per_make)
```

---
**6. For each fuel type, find the minimum and maximum engine size. This gives insight into what range of engine sizes each fuel category typically covers.**

```python
# Write your code here

engine_size_range = data.groupby('Fuel Type')['Engine Size(L)'].agg(['min','max'])

print(engine_size_range)
```

```python
# Write your code here to create a visualization
import matplotlib.pyplot as plt

# Bar chart for min and max engine sizes by fuel type
engine_size_range.plot(kind='bar',
                       figsize=(10, 6),
                       rot=45)

plt.title('Min and Max Engine Size (L) by Fuel Type')
plt.ylabel('Engine Size (L)')
plt.xlabel('Fuel Type')
plt.legend(title='Engine Size Stats')
plt.tight_layout()
plt.show()

```

---
**7. Display the top fuel-efficient vehicle for each fuel type in descending order based on the highest combined fuel consumption (mpg).**



```python
best_fuel_efficiency_per_fuel = data.loc[data.groupby("Fuel Type")["Fuel Consumption Comb (mpg)"].idxmax(),
                                         ["Fuel Type", "Fuel Consumption Comb (mpg)"]]

# Sort by fuel efficiency in descending order
best_fuel_efficiency_per_fuel = best_fuel_efficiency_per_fuel.sort_values(by="Fuel Consumption Comb (mpg)", ascending=False)

print(best_fuel_efficiency_per_fuel)
```
